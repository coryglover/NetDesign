```python
robotO = {}
robottypes = ["L","Y","J","E","C","M","U2","U4","W"]
for n in robottypes:
    robotO[n] = {}
    for m in robottypes:
        robotO[n][m] = 0



robotO["L"]["J"] = 2
robotO["L"]["E"] = 1
robotO["L"]["W"] = 1
robotO["Y"]["U2"] = 2
robotO["Y"]["U4"] = 4
robotO["J"]["L"] = 2
robotO["J"]["M"] = 1
robotO["E"]["L"] = 1
robotO["C"]["M"] = 1
robotO["C"]["U2"] = 1
robotO["C"]["U4"] = 1
robotO["M"]["J"] = 1
robotO["M"]["C"] = 1
robotO["U2"]["Y"] = 2
robotO["U2"]["E"] = 2
robotO["U2"]["C"] = 2
robotO["U4"]["Y"] = 2
robotO["U4"]["E"] = 2
robotO["U4"]["C"] = 4
robotO["W"]["L"] = 1

robotc = {"L":2,"Y":2,"J":2,"E":1,"C":2,"M":2,"U2":2,"U4":4,"W":1}
```

```python
graphs = os.listdir('../data/robots/bl_4/edgefiles/')
Xfiles = [f'X_{g[:-5]}.txt' for g in graphs]
robot_path = '../data/robots/bl_4/'
for i in range(len(graphs)):
    g = nx.read_edgelist(robot_path + 'edgefiles/' + graphs[i])
    with open(robot_path + 'Xfiles/' + Xfiles[i],'r') as f:
        X_labels = f.read().splitlines()[0]
    X_labels = X_labels.split(',')[:-1]
    X_labels[0] = X_labels[0][-1]
    O = np.zeros((len(X_labels),len(X_labels)),dtype=int)
    for l,key1 in enumerate(X_labels):
        for m,key2 in enumerate(X_labels):
            O[l,m] = robotO[key1][key2]

    c = np.zeros(len(X_labels),dtype=int)
    for l,key in enumerate(X_labels):
        c[l] = robotc[key]

    np.savetxt(robot_path + 'Ofiles/O_' + graphs[i][:-5] + '.txt', O)
    np.savetxt(robot_path + 'cfiles/c_' + graphs[i][:-5] + '.txt', c)
```
